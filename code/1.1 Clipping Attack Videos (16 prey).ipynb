### Creating Images from the Attack Videos

```python
import os

from google.colab import drive
drive.mount('/content/drive')
```

```python
attacks_dir = "/content/drive/MyDrive/Predator Prey Thesis/Data/1. Data Processing/Raw/pred_attack/pred_attack_16"

images_dir = os.path.join(attacks_dir, "images")
os.makedirs(images_dir, exist_ok=True)
```

```python
num_attacks = 27

for i in range(1, num_attacks + 1):
    video_path = os.path.join(attacks_dir, f"pred_attack_{i}_16.mp4")
    if not os.path.exists(video_path):
        print(f"Skipping attack {i}, file not found: {video_path}")
        continue

    out_folder = os.path.join(images_dir, f"pred_attack_{i}_16")
    os.makedirs(out_folder, exist_ok=True)

    # temporary extraction folder
    temp_folder = os.path.join(out_folder, "temp")
    os.makedirs(temp_folder, exist_ok=True)

    command = (
        f'ffmpeg -i "{video_path}" '
        f'-vf "select=not(mod(n\\,2))" '
        f'-vsync vfr '
        f'"{temp_folder}/frame_%03d.jpg"'
    )

    os.system(command)

    # rename with original frame numbers (0,2,4,6...)
    files = sorted(os.listdir(temp_folder))

    for idx, fname in enumerate(files):
        old_path = os.path.join(temp_folder, fname)

        original_frame = idx * 2
        new_name = f"frame_{original_frame:03d}.jpg"

        new_path = os.path.join(out_folder, new_name)
        os.rename(old_path, new_path)

    # remove temp folder
    os.rmdir(temp_folder)

    print(f"Attack {i}: extracted {len(files)} frames -> {out_folder}")
```

Checking how many frames are in each 5 second video.

```python
import cv2

video_dir = "/content/drive/MyDrive/Predator Prey Thesis/Data/1. Data Processing/Raw/pred_attack/pred_attack_16"

for i in range(1, 28):
    video_path = os.path.join(video_dir, f"pred_attack_{i}_16.mp4")

    if not os.path.exists(video_path):
        continue

    cap = cv2.VideoCapture(video_path)

    frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))
    fps = cap.get(cv2.CAP_PROP_FPS)
    duration = frames / fps

    cap.release()

    print(
        f"Attack {i}: "
        f"{frames} frames | "
        f"{fps:.2f} FPS | "
        f"{duration:.2f} seconds"
    )
```
